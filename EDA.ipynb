```python
import matplotlib.pyplot as plt

plot_df = df_breweries_clean[["longitude", "latitude"]].dropna()

plt.figure(figsize=(8, 6))
plt.scatter(plot_df["longitude"], plot_df["latitude"], alpha=0.5)
plt.show()
```

```python
subset_cols = ["name", "address_1", "city", "country"]
mask = df_breweries_clean.duplicated(subset=subset_cols, keep=False)
dup_addr = df_breweries_clean.loc[mask, :].sort_values(subset_cols)
print(f"\nMögliche Duplikate (gleicher Name + Adresse + Stadt + Land): "
      f"{dup_addr.shape[0]} Zeilen "
      f"({dup_addr.duplicated(subset=subset_cols).sum()} echte Duplikate)")

display(dup_addr)
```

```python
df_breweries_clean = df_breweries_clean.drop_duplicates(
    subset=["name", "address_1", "city", "country"],
    keep="first"
)
```

```python
subset_cols = ["name", "address_1", "city", "country"]
mask = df_breweries_clean.duplicated(subset=subset_cols, keep=False)
dup_addr = df_breweries_clean.loc[mask, :].sort_values(subset_cols)
print(f"\nMögliche Duplikate (gleicher Name + Adresse + Stadt + Land): "
      f"{dup_addr.shape[0]} Zeilen "
      f"({dup_addr.duplicated(subset=subset_cols).sum()} echte Duplikate)")

display(dup_addr)
```

```python
from pathlib import Path

df = df_breweries_clean
save_path = Path("figs")

# Brewery Type
type_counts = df.loc[:, "brewery_type"].value_counts()

fig, ax = plt.subplots(figsize=(8, 5))
type_counts.plot(kind="bar", ax=ax, color="#4c72b0")
ax.set_title("Verteilung Brewery Type")
ax.set_xlabel("Brewery Type")
ax.set_ylabel("Anzahl")
plt.tight_layout()
plt.savefig(save_path / "plot_brewery_type.png", dpi=150)
plt.close()

# 2. Top-10 Länder
country_counts = df.loc[:, "country"].value_counts().head(10)

fig, ax = plt.subplots(figsize=(8, 5))
country_counts.plot(kind="bar", ax=ax, color="#55a868")
ax.set_title("Top 10 Länder nach Anzahl Brauereien")
ax.set_xlabel("Land")
ax.set_ylabel("Anzahl")
plt.tight_layout()
plt.savefig(save_path / " plot_top10_countries.png", dpi=150)
plt.close()

# 3. Top-10 Bundesstaaten/Regionen (gesamt, über alle Länder)
state_counts = df.loc[:, "state_province"].value_counts().head(10)

fig, ax = plt.subplots(figsize=(8, 5))
state_counts.plot(kind="bar", ax=ax, color="#c44e52")
ax.set_title("Top 10 Bundesstaaten/Regionen")
ax.set_xlabel("Region")
ax.set_ylabel("Anzahl")
plt.tight_layout()
plt.savefig(save_path / "plot_top10_states.png", dpi=150)
plt.close()

# 4. Bonus: brewery_type-Verteilung nur für USA
us_type_counts = df.loc[df.loc[:, "country"] == "United States", "brewery_type"].value_counts()
fig, ax = plt.subplots(figsize=(8, 5))
us_type_counts.plot(kind="bar", ax=ax, color="#8172b2")
ax.set_title("Verteilung Brewery Type in den USA")
ax.set_xlabel("Brewery Type")
ax.set_ylabel("Anzahl")
plt.tight_layout()
plt.savefig(save_path / "plot_brewery_type_usa.png", dpi=150)
plt.close()
```
